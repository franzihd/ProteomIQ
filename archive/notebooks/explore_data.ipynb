```python
import pandas as pd

proteomics = pd.read_csv('/home/franziska/bachelor/ProteomIQ/data/luad/proteomics_raw.csv', header=[0,1], index_col=0)
clinical = pd.read_csv('/home/franziska/bachelor/ProteomIQ/data/luad/clinical.csv', index_col=0)

both = proteomics.index.intersection(clinical.index)

print(f"Proteomics samples: {len(proteomics)}")
print(f"Clinical samples: {len(clinical)}")
print(f"Samples in both datasets: {len(both)}")

print(f"Total rows: {len(proteomics)}")
print(f"Unique patient IDs: {len(proteomics.index.unique())}")
```
